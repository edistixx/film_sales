```
# Import data
!wget https://cdn.freecodecamp.org/project-data/health-costs/insurance.csv
dataset = pd.read_csv('insurance.csv')
dataset.tail()
```

```
for i in dataset.region.unique():
  if i == 'southeast':
    dataset.loc[dataset.region == i,"region"] = 1
  if i == 'southwest':
    dataset.loc[dataset.region == i,"region"] = 2
  if i == 'northeast':
    dataset.loc[dataset.region == i,"region"] = 3
  if i == 'northwest':
    dataset.loc[dataset.region == i,"region"] = 4
for u in dataset.sex.unique():
  if u == 'male':
    dataset.loc[dataset.sex == u,"sex"] = 1
  if u == 'female':
    dataset.loc[dataset.sex == u,"sex"] = 0



for z in dataset.smoker.unique():
  if z == 'yes':
    dataset.loc[dataset.smoker == z,"smoker"] = 1
  if z == 'no':
    dataset.loc[dataset.smoker == z,"smoker"] = 0

# from sklearn.preprocessing import StandardScaler
# scaler = StandardScaler()
# scaler = scaler.fit(dataset)
# data = scaler.transform(dataset)


dataset.age = dataset.age.astype("float")
dataset.sex = dataset.sex.astype("float")
dataset.children = dataset.children.astype("float")
dataset.smoker = dataset.smoker.astype("float")
dataset.region = dataset.region.astype("float")


X = dataset.drop("expenses",axis = 1)
y = dataset["expenses"]



train_dataset, test_dataset, train_labels, test_labels = train_test_split(X, y, test_size=0.2)
# a = [train_dataset, test_dataset, train_labels, test_labels]
# [i.shape for i in a]
# # dataset.corr()
train_dataset

```

```
from keras.models import Sequential
from keras.layers import Dense
model = Sequential()
model.add(Dense(128,input_dim = train_dataset.shape[1],activation = "relu"))
model.add(Dense(64,activation = "relu"))
model.add(Dense(32,activation = "relu"))
model.add(Dense(16,activation = "relu"))
model.add(Dense(1,activation = "linear"))

model.compile(loss = "mean_squared_error",optimizer = "adam",metrics = ["mae", "mse" ])
model.summary()
history = model.fit(train_dataset.values,train_labels.values,epochs = 100,batch_size = 2,validation_split = 0.2)
```

```
pred = model.predict(test_dataset).flatten()
l = test_labels.values.flatten()
data1 = pd.DataFrame({"act":l,"pred":pred,"err":(np.abs(l - pred)/(l))*100})
# history.history['val_loss']
test_dataset = test_dataset.values
test_labels = test_labels.values
```

```
# RUN THIS CELL TO TEST YOUR MODEL. DO NOT MODIFY CONTENTS.
# Test model by checking how well the model generalizes using the test set.
loss, mae, mse = model.evaluate(test_dataset, test_labels, verbose=2)

print("Testing set Mean Abs Error: {:5.2f} expenses".format(mae))

if mae < 3500:
  print("You passed the challenge. Great job!")
else:
  print("The Mean Abs Error must be less than 3500. Keep trying.")

# Plot predictions.
test_predictions = model.predict(test_dataset).flatten()

a = plt.axes(aspect='equal')
plt.scatter(test_labels, test_predictions)
plt.xlabel('True values (expenses)')
plt.ylabel('Predictions (expenses)')
lims = [0, 50000]
plt.xlim(lims)
plt.ylim(lims)
_ = plt.plot(lims,lims)
```
